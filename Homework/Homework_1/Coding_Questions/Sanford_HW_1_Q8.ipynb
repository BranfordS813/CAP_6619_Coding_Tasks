## Q8.1

### Importing housing.header.txt as dataframe

```python
# Import dataset

housing = pd.read_csv('housing.header.txt', sep=',') 
```

```python
housing.head()
```

### Number of features and instances

```python
# Number of features corresponds to columns of dataset
# Number of instances corresponds to rows of dataset

num_features = len(housing.columns);
num_instances = len(housing.index); 

print(f"There are {num_features} features and {num_instances} instances within the housing.header.txt dataset")
```

### Plotting Crim feature over all instances

```python
# As based on Python.Simple.Analysis.html

# Get y value of crim
y_crim = housing['Crim'];

# Get instance number
x_crim = y_crim.shape[0];
print(f"Number of instances associated with crim: {x_crim}")

# Index x_crim
x_index = np.arange(x_crim)
# print(x_index)

plt.scatter(x_index, y_crim, marker = 'o', alpha=0.7) # Alpha is added to help discern overlaping datapoints
plt.xlabel('Instance Index (sample)')
plt.ylabel('Crim Value')
plt.title('Crim Feature Over Sample Instances')
plt.grid(True)
plt.show
```

## Q8.2

Scatter plots for crim, rm, age, and tax features paired with medv

```python
# set features as variables

# Medv as y-value
y_medv = housing['Medv']

# Crim, Rm, Age, and Tax as x-values 
x_crim = housing['Crim']
x_rm = housing['Rm']
x_age = housing['Age']
x_tax = housing['Tax']

# Plotting colums as 1x4 frame
plt.figure(figsize=(20, 4))

# plot 1: Crim vs Medv
plt.subplot(1, 4, 1)
plt.scatter(x_crim, y_medv, alpha=0.6)
plt.xlabel('Crim')
plt.ylabel('Medv')
plt.title('Crim vs Medv')

# plot 2: Rm vs Medv
plt.subplot(1, 4, 2)
plt.scatter(x_rm, y_medv, alpha=0.6)
plt.xlabel('Rm')
plt.ylabel('Medv')
plt.title('Rm vs Medv')

# plot 3: Age vs Medv
plt.subplot(1, 4, 3)
plt.scatter(x_age, y_medv, alpha=0.6)
plt.xlabel('Age')
plt.ylabel('Medv')
plt.title('Age vs Medv')

# plot 4: Tax vs Medv
plt.subplot(1, 4, 4)
plt.scatter(x_tax, y_medv, alpha = 0.6)
plt.xlabel('Tax')
plt.ylabel('Medv')
plt.title('Tax vs Medv')

plt.tight_layout()
plt.show()
```

From the plots, it is clear that the Median Value of the house (MedV) is sensitive to key features like level of crime in the area (Crim), number of rooms (rm) the age of the house (age) and taxes based on proper value. Most of these patterns are intuitive

- The lower the crime right, the higher the value of the house 
- The more rooms the house possess, the higher the hosues value. Some hoses, outside of outliers, can have high prices with either a small or medium amount of room, perhaps due to other factors of the house, like size or accomodations
- All though not exactly clear, it appears that the correlation between house value and age is that the newer the house, the higher it's value, though it skewes towards a flattened relatioship where on average, at least for houses 40 to 80 years old, the value remains around 30K to 20K
- The Tax vs Medv chart is intersting as the data looks more akin to a box and whisker plot where it apperas that medians with quartiles are present. still, based on the pattern, the lower the proper taxes of the area, the higher the value of the house.

## Q8.6

Pairwise correlation between every two variables - the correslations can either be displayed as a matrix or level plot

Displaying Correlations as a matrix:
A pairwise correlation matrix can be used. The seaborn python package can be used to visualize this.

```python
# Pairwise Correlation Matrix
corr_matrix = housing.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', center=0)
plt.title('Pairwise Correlation Matrix for Housing Dataset')
plt.show()
```

## Q8.7
Feature most correlated to Medv

```python
# The corrleated numbers from the seaborn matrix can be sorted in descending order (from higest to least);
# corresonding with the median value.
medv_corr_sorted = housing.corr()['Medv'].sort_values(ascending=False)
print(medv_corr_sorted)
```

Rm (Room number) is the most correlated with Medv, meaning that the higher the room number, by correlation, the higher the value of the home. of note, the second most correlated is Zn, the number of residential land zoned lots (over 25,000 square feet).

The most negatively correlated feature is Lstat - the percentage of low socioeconomic status in the population; the second lowest was Ptratio, pupil to teacher ratio by town. As these are negative correlations, they show that as they go up, the prices of homes go down. for Ptratio, it is more indicative or more students present than teachers (i.e., larger class sizes).

## Q8.9

Adding a new instance to the dataset

```python
# A dictionary can be used to create the new instance, which can then 
# be concatenated to the dataframe via pandas

# Define new row (instance) as a dictionary 
new_instance = {
    'Crim': 1.0,
    'Zn': 0.2,
    'Indus': 6,
    'Chas': 0.1,
    'Nox': 6.5,
    'Rm': 5,
    'Age': 100,
    'Dis': 4.1,
    'Rad': 4.5,
    'Tax': 21,
    'Ptratio': 20,
    'B': 300,
    'Lstat': 12,
    'Medv': 20.5
}

# Convert new instance into dataframe
new_instance_df = pd.DataFrame([new_instance])

# Concatenate new insntace into the dataframe as a new dataframe

new_housing = pd.concat([housing, new_instance_df], ignore_index=True)

num_new_instances = len(new_housing.index)

print(f"The number of instances of the old housing dataset was {num_instances} instances")
print(f"The number of instnaces of the new housing dataset is now {num_new_instances} instances")
```
