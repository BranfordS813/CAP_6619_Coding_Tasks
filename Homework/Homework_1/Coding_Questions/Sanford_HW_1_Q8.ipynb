```python
# Import libraries

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

## Q8.1

### Importing housing.header.txt as dataframe

```python
# Import dataset

housing = pd.read_csv('housing.header.txt', sep=',') 
```

```python
housing.head()
```

### Number of features and instances

```python
# Number of features corresponds to columns of dataset
# Number of instances corresponds to rows of dataset

features = len(housing.columns);
instances = len(housing.index); 

print(f"There are {features} features and {instances} instances within the housing.header.txt dataset")
```

### Plotting Crim feature over all instances

```python
# As based on Python.Simple.Analysis.html

# Get y value of crim
y_crim = housing['Crim'];

# Get instance number
x_crim = y_crim.shape[0];
print(f"Number of instances associated with crim: {x_crim}")

# Index x_crim
x_index = np.arange(x_crim)
# print(x_index)

plt.scatter(x_index, y_crim, marker = 'o')
plt.xlabel('Instance Index (sample)')
plt.ylabel('Crim Value')
plt.title('Crim Feature Over Sample Instances')
plt.grid(True)
plt.show
```

## Q8.2

### Paring features from dataset

```python
# Sorting paired datasets

# crim_medv
crim_medv = housing[['Crim','Medv']]
crim_medv_sort = crim_medv.sort_values('Crim')

# rm_medv
rm_medv = housing[['Rm', 'Medv']]
rm_medv_sort = rm_medv.sort_values('Rm')

# age_medv
age_medv = housing[['Age', 'Medv']]
age_medv_sort = age_medv.sort_values('Age')

# tax_medv
tax_medv = housing[['Tax', 'Medv']]
tax_medv_sort = tax_medv.sort_values('Tax')

```

```python
# Scatter sub plot correlations with medv

# y-values for plots
CM_y_medv = crim_medv_sort['Medv'];
RM_y_medv = rm_medv_sort['Medv'];
AGE_y_mdev = age_medv_sort['Medv'];
TAX_y_medv = tax_medv_sort['Medv'];

# x-values for plots
x_crim = crim_medv_sort['Crim'];
x_rm = rm_medv_sort['Rm'];
x_age = age_medv_sort['Age'];
x_tax = tax_medv_sort['Tax'];

# Subplots 

# plt.figure(figsize=(20,5))
plt.subplot(2,2,1)
plt.scatter(x_crim, CM_y_medv, marker = 'o')
plt.xlabel('Crim')
plt.ylabel('Medv')
plt.title('Crim - Medv')

# plt.figure(figsize = (20, 5))
plt.subplot(2,2,2)
plt.scatter(x_rm, RM_y_medv, marker = 'o')
plt.xlabel('Rm')
plt.ylabel('Medv')
plt.title('Rm - Medv')

# plt.figure(figsize = (20, 5))
plt.subplot(2,2,3)
plt.scatter(x_age, AGE_y_mdev, marker = 'o')
plt.xlabel('Age')
plt.ylabel('Medv')
plt.title('Age - Medv')

# plt.figure(figsize = (20, 5))
plt.subplot(2,2,4)
plt.scatter(x_tax, TAX_y_medv, marker = 'o')
plt.xlabel('Tax')
plt.ylabel('Medv')
plt.title('Tax - Medv')
```

From the plots, it appears that the median value for housing varies heavily with factors like age, amount of rooms, crime rate, and tax. The less crime in the area, the higher the value; the more rms in the house, the higher the value; the lower the property-tax rate, the higher the value; and interstingly enough, finding a correlation between age of the home owners in the houses (before 1940) shows a sgrong coreelation with older individuals with median values; although median values are shown throughout, with soem outliers for most ages.

```python
# Following feature correlation plot from 
# CAP 6619.Python.Simple.Analysis.html

# Shows correlation as a heatmap

# feature correlation plot

plt.matshow(boston.corr(),fignum=1,aspect='auto')
plt.xticks(range(len(boston.columns)), boston.columns)
plt.yticks(range(len(boston.columns)), boston.columns)
plt.colorbar()
plt.show()
```

```python
# Seaborn can also be used like the matshow() metehod tied in
# with corr():

annot = True # enable correlation value annotation
cmap = 'coolwarm';
center = 0;

sns.heatmap(boston.corr(), annot=annot, cmap=cmap, center=center)
plt.title('Boston Df Correlation Matrix')
plt.show()


```

From the heatmap, it appears that the Rm attribute is most correlated to Medv, showing that the market value for houses goes up with higher room numbers.
